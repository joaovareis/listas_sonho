```python
import matplotlib.pyplot as plt
import pandas as pd

raw = pd.read_csv('imu_data_original.csv')
fil = pd.read_csv('imu_data_filtrado.csv')
raw.columns = raw.columns.str.strip().str.replace('#', '')
fil.columns = fil.columns.str.strip().str.replace('#', '')

plt.figure(figsize=(12, 6))

plt.plot(raw['timestamp'], raw['acel_linear'], label='IMU sem filtro', color='gray')

plt.plot(fil['timestamp'], fil['acel_linear'], label='IMU filtrada', color='blue')

plt.title('Comparação entre sinal original e filtrado por média móvel', fontsize=14)
plt.xlabel('Timestamp', fontsize=12)
plt.ylabel('Magnitude (m/s^2)', fontsize=12)

plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(loc='upper right', fontsize=11)

plt.show()
```
