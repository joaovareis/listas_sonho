```python
import matplotlib.pyplot as plt
import pandas as pd

csv_data = pd.read_csv('data/output/relatorio_robo.csv', names=['sensor', 'timestamp_index', 'metrica', 'valor'])

sensores = csv_data['sensor'].unique()

fig, grafs = plt.subplots(nrows=3, ncols=1, figsize=(10, 12))

for i in range(3):
    nome = sensores[i]
    area = grafs[i]

    csv_data_sensor = csv_data[csv_data['sensor'] == nome]
    
    for metrica in csv_data_sensor['metrica'].unique():
        csv_data_metrica = csv_data_sensor[csv_data_sensor['metrica'] == metrica]
        area.plot(csv_data_metrica['timestamp_index'], csv_data_metrica['valor'], label=metrica)
    
    area.set_title(f'Sensor: {nome}')
    area.set_xlabel('Timestamp / Índice')
    area.set_ylabel('Valor')
    area.legend(loc='upper right')
    area.grid(True, alpha=0.6)

plt.tight_layout()
plt.show()
```
