Exercício 1

```python
import matplotlib.pyplot as plt
import numpy as np

NUM_AMOSTRAS = 100 #10 Hz de amostras

time = np.linspace(0, 10, NUM_AMOSTRAS)
ruido = np.random.normal(0, 0.3, NUM_AMOSTRAS)

ideal = np.sin(time) + 25

real = ideal + ruido

plt.plot(time, ideal, color='blue')
plt.plot(time, real, ".", color='red')

plt.title("Simulação de sensor de temperatura")

plt.xlabel('Tempo (s)')
plt.ylabel('Temperatura (°C)')

plt.grid()

plt.show()
```

Exercício 2

```python
rng = np.random.default_rng()

NUM_JOGADAS = 1000

dado1 = rng.integers(1, 7, NUM_JOGADAS)
dado2 = rng.integers(1, 7, NUM_JOGADAS)

mult_dados = dado1 * dado2

plt.hist(mult_dados, bins=np.arange(1.5, 37.5, 1), edgecolor='gray', color='lightgreen')

plt.title(f"Distribuição da multiplicação de dois dados em {NUM_JOGADAS} jogadas")

plt.xlabel('Valor resultante')
plt.ylabel('Frequência')

plt.grid(axis='y')

plt.xticks(np.arange(0, 37, 2))

plt.show()
```

Exercício 3

```python
import math

points_world = np.array([
[2, 1],
[4, 3],
[1, 5],
[6, 2]
])

robot_position = np.array([3, 2])
robot_orientation = np.pi/4

transladada = points_world - robot_position

matriz_rotacao = np.array([
    [math.cos(-robot_orientation), -math.sin(-robot_orientation)],
    [math.sin(-robot_orientation),  math.cos(-robot_orientation)]
])

rotacionada = [matriz_rotacao @ point for point in transladada]

for point in points_world: plt.scatter(point[0], point[1], color='blue')

plt.scatter(robot_position[0], robot_position[1], color='green')

for point in rotacionada: plt.scatter(point[0], point[1], color='red')

plt.title(f"Posição de pontos de interesse no frame world (azul) e no frame body (vermelho)")

plt.xlabel('x')
plt.ylabel('y')

plt.xticks(np.arange(-2, 7, 1))
plt.yticks(np.arange(-2, 7, 1))

plt.grid()
plt.show()
```
